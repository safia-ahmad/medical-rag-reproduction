```python
from datasets import load_dataset

# Load the precomputed embeddings dataset (if not already loaded)
# This is added to resolve 'NameError: name 'embeddings' is not defined'
embeddings = load_dataset("slinusc/PubMedAbstractsSubsetEmbedded")

# Take a manageable subset (50k docs) to work with first
# This is added to resolve 'NameError: name 'subset' is not defined'
subset = embeddings['train'].select(range(50000))

!pip install -q rank_bm25

from rank_bm25 import BM25Okapi

# Tokenize each abstract into words (BM25 works on word overlap, so it needs simple word lists)
tokenized_corpus = [doc.lower().split() for doc in subset['abstract']]

bm25 = BM25Okapi(tokenized_corpus)

print(f"BM25 index built on {len(tokenized_corpus)} documents")
```

```python
question = "What are the effects of low-molecular weight substances on tumor growth?"

# BM25 search
tokenized_query = question.lower().split()
bm25_scores = bm25.get_scores(tokenized_query)
top_bm25_idx = np.argsort(bm25_scores)[::-1][:5]

print("Top 5 BM25 results:")
for idx in top_bm25_idx:
    # Convert numpy.int64 to standard Python int for indexing the dataset
    print(f"  Score {bm25_scores[idx]:.2f} — {subset['title'][int(idx)][:80]}")
```

```python
!pip install -q transformers torch

from transformers import AutoTokenizer, AutoModel
import torch
import faiss
import numpy as np

# Load MedCPT's query encoder (a different model than the document encoder, by design)
tokenizer = AutoTokenizer.from_pretrained("ncbi/MedCPT-Query-Encoder")
model = AutoModel.from_pretrained("ncbi/MedCPT-Query-Encoder")

def embed_query(text):
    inputs = tokenizer(text, truncation=True, padding=True, return_tensors="pt", max_length=64)
    with torch.no_grad():
        embeds = model(**inputs).last_hidden_state[:, 0, :]
    return embeds.numpy().astype('float32')

# Build the FAISS index from the embedding vectors
# Added to resolve 'NameError: name 'index' is not defined'
embedding_matrix = np.array(subset['embedding'], dtype='float32')
dimension = embedding_matrix.shape[1]
index = faiss.IndexFlatIP(dimension)
index.add(embedding_matrix)

# Embed our question and search FAISS
query_vec = embed_query(question)
distances, faiss_idx = index.search(query_vec, 5)

print("Top 5 MedCPT/FAISS results:")
for idx in faiss_idx[0]:
    print(f"  {subset['title'][int(idx)][:80]}")
```

```python
def hybrid_search(question, top_k_bm25=50, top_k_final=5):
    # Step 1: BM25 narrows down candidates fast
    tokenized_query = question.lower().split()
    bm25_scores = bm25.get_scores(tokenized_query)
    top_bm25_idx = np.argsort(bm25_scores)[::-1][:top_k_bm25]

    # Step 2: MedCPT reranks just those candidates by meaning
    candidate_embeds = embedding_matrix[top_bm25_idx]
    query_vec = embed_query(question)
    scores = candidate_embeds @ query_vec.T  # similarity scores
    reranked = top_bm25_idx[np.argsort(scores.flatten())[::-1][:top_k_final]]

    return reranked

results = hybrid_search(question)
print("Top 5 Hybrid (BM25→MedCPT) results:")
for idx in results:
    print(f"  {subset['title'][int(idx)][:80]}")
```
